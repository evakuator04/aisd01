```python
#1
def bi(list,key):
  if len(list)==0: #Если в списке нет элементов, пишем, что он пустой
    return 'Пустой массив'
  low=0 #Левая граница поиска внутри списка
  high= len(list)-1 #Правая граница поиска внутри списка
  por= list[0]<list[-1] #Проверяем, как идут числа: по возрастанию или по убыванию
  while low<=high:
    mid=low+(high-low)//2 #Находим середину отрезка
    midl = list[mid] #Берем значение элемента из середины массива
    if midl==key: #Если нашли нужное число, возвращаем его индекс
      return mid
    if por: #Если массив возрастает
      if midl>key: #Если центральный элемент больше искомого, сдвигаем правую границу левее
        high=mid-1
      elif midl<key: #Иначе сдвигаем левую границу правее
        low=mid+1
    else: #Если массив убывает
      if midl>key: #Если центральный элемент больше искомого, сдвигаем левую границу правее
        low=mid+1
      elif midl<key: #Иначе сдвигаем правую границу левее
        high=mid-1
  return low
a=[1,2,3,5,7,8]
b=[]
c=[1]
d=[10,7,3,2,1]
print(bi(a,3),bi(a,6))
print(bi(b,3))
print(bi(c,1),bi(c,4))
print(bi(d,3),bi(d,8))
```

```python
#2
def gor(arr):
  if len(arr)<3: #Проверяем, что длина массив может быть горным
    return 'Массив не горный'
  low=0 #Левая граница поиска внутри списка
  high=len(arr)-1 #Правая граница поиска внутри списка
  while low < high:
        mid = low + (high - low) // 2 #Находим середину отрезка
        if arr[mid] > arr[mid + 1]: #Если текущий элемент больше следующего, значит, мы на убывающем склоне
            high = mid #Сдвигаем правую границу в середину тк вершина либо в ней, либо левее
        else: #Если текущий элемент больше следующего, значит, мы на возрастающем склоне
            low = mid + 1 #Сдвигаем левую границу за середину тк середина точно не вершина

  peak = low #Когда границы сошлись low и high обозначают вершину
  if peak == 0 or peak == len(arr) - 1: #Проверяем, что вершина не первый элемент и не последний
        return 'Массив не горный'

  if arr[peak - 1] < arr[peak] > arr[peak + 1]: #Проверяем условие вершины
        return peak

a=[1,2,3,4,5,2,1]
b=[1,1,1]
c=[]
d=[2]
e=[1,2,3,4,4,4]
print(gor(a))
print(gor(b))
print(gor(c))
print(gor(d))
print(gor(e))
```

```python
#3
def ch(lis):
    if len(lis)==0: #Если в списке нет элементов, пишем, что он пустой
        return "Массив пустой"

    low=0 #Левая граница поиска внутри списка
    high = len(lis)-1 #Правая граница поиска внутри списка
    noneg = len(lis) #Переменная для индекса первого неотрицательного числа
    while low <= high: #Бинарный поиск границы отрицательных чисел
        mid = low+(high-low)//2 #Находим середину отрезка
        if lis[mid] >= 0: #Если элемент в середине неотрицательный
            noneg = mid #Запоминаем этот индекс как возможную границу
            high = mid - 1 #Сдвигаем правую границу левее, чтобы найти более ранний ноль или положительное число
        else:
            low = mid + 1 #Если элемент отрицательный, ищем в правой половине


    neg_count = noneg #Индекс первого неотрицательного числа равен количеству отрицательных чисел

    low=0 #Левая граница поиска внутри списка
    high = len(lis)-1 #Правая граница поиска внутри списка
    posit = len(lis) #Переменная для индекса первого положительного числа

    while low <= high: #Бинарный поиск границы положительных чисел
        mid = low+(high-low)//2 #Находим середину отрезка
        if lis[mid] > 0: #Если элемент в середине положительный
            posit = mid #Запоминаем этот индекс как начало положительных чисел
            high = mid-1 #Сдвигаем правую границу левее, чтобы найти первое положительное число
        else:
            low = mid+1 #Если элемент меньше или равен нулю, то сдвигаем левую границу правее

    pos_count = len(lis) - posit #Количество положительных чисел — это длина всего массива минус индекс первого положительного числа

    if pos_count > neg_count:
        return 'Положительных чисел больше', pos_count
    elif neg_count > pos_count:
        return 'Отрицательных чисел больше', neg_count
    else:
        return 'Поровну', pos_count

a = [-5, -3, -1, 0, 0, 2, 4, 6, 8]
b=[-2,-1,1]
c=[]
d=[1,1,1,1]
print(ch(a))
print(ch(b))
print(ch(c))
print(ch(d))
```

```python
#4
def small(nums):
    if len(nums) == 0:
        return []

    counts = [0] * len(nums)
    seen = [] #В этот список мы будем складывать уже просмотренные числа по порядку (от меньшего к большему)

    for i in range(len(nums) - 1, -1, -1): # Идем по нашему массиву с конца в начало (справа налево)
        nownum = nums[i]
        low = 0  #Левая граница поиска внутри списка seen
        high = len(seen) - 1  #Правая граница поиска внутри списка seen



        indx = len(seen) #По умолчанию считаем, что число самое большое и встанет в самый конец

        #Пока левая граница не пересечет правую
        while low <= high:
            mid = (low + high) // 2  #Находим середину отрезка


            if seen[mid] >= nownum: #Если число в середине больше или равно нашему текущему числу
                indx = mid  #Запоминаем это место (наше число должно встать левее)
                high = mid - 1  #Сдвигаем правую границу левее, ищем дальше в левой половине
            else:
                low = mid + 1  #Если число в середине меньше нашего, ищем в правой половине

        counts[i] = indx #Индекс, который нашел бинарный поиск (indx), равен количеству чисел, которые меньше нашего текущего числа

        seen.insert(indx, nownum) #Вставляем наше число на правильное место в отсортированный список seen, чтобы он оставался упорядоченным для следующих шагов

    return counts
nums = [5, 2, 6, 1]
result = small(nums)

print(result)
```
