```python
import math
def f(x):
  return math.sqrt(1-x)-math.tan(x)
L=0 #левая граница
R=1 #правая граница
eps=0.0001 #точность поиска корня
step=0 #подсчет шагов
while abs(R-L)>eps: #пока отрезок больше заданной точности
  step+=1
  B=(L+R)/2 #находим середину текущего отрезка
  print(step,'.',round(B,4)) #номер шага и текущее значение
  if f(L)*f(B)<=0: #проверяем в какой половине находится корень сравнивая с нулем и меняем границу в зависимости от половины
    R=B
  else:
    L=B
print(round(B,4)) #для 13 варианта

```

```python
import math
def f(x):
  return math.sqrt(1-x)-math.tan(x)
L=0 #левая граница
R=1 #правая граница
eps=0.0001 #точность поиска корня
step=0 #подсчет шагов
x0=L #переменная для хранения значения x с предыдущего шага
while True: #бесконечный цикл, который прервется при достижении заданной точности
  step+=1
  x=L-(f(L)*(R-L))/(f(R)-f(L))#формула метода хорд
  if step>1 and abs(x-x0)<=eps: #если разность между текущим х и прошлым меньше заданной точности то прерываем цикл
    break
  x0=x
  if f(x)==0: #если вдруг функция в х будет равна нулю то выводим корень
    print(round(x,4))
  if f(x)*f(R)>0: #если знаки в х и R одинаковы, то сдвигаем правую границу в х
    R=x
  else:
    L=x #если разные то сдвигаем левую границу в х
print(round(x,4)) #для 13 варианта
```
